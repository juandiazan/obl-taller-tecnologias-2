```python
import pandas as pd
from langchain_community.document_loaders import PyPDFLoader
from sentence_transformers import SentenceTransformer
```

# Cargar dataset de CSV y quitar columnas que no agreguen información/poco interesantes

```python
# https://www.kaggle.com/datasets/fronkongames/steam-games-dataset

df = pd.read_csv('content/games.csv')

df = df.drop(columns=['AppID', 'Screenshots', 'Movies', 'Header image', 'Support url',
                      'Metacritic url', 'User score', 'Score rank',
                      'Average playtime two weeks', 'Median playtime two weeks'])

display(df.head())
```

```python
model = SentenceTransformer("all-MiniLM-L6-v2")
```

```python
rows = []
for _, row in df.iterrows():
    parts = [f"{col}: {row[col]}" for col in df.columns if pd.notna(row[col])]
    rows.append(", ".join(parts))
```

```python
print("First element: ", rows[0])
print("Second element:", rows[1])
```

```python
rowsToEncode = []
# Slice the list to process the first 10% of the records
for row in rows[:len(rows)//10]:
    rowsToEncode.append(row)

print(len(rowsToEncode))
```

```python
model.encode(rowsToEncode)
```

# Cargar papers dados en Aulas

```python
documents = []
routes = ["2507.14805v1.pdf", "Attention is all you need.pdf",
          "Designing Data-Intensive Applications The Big Ideas Behind Reliable, Scalable, and Maintainable Systems by Martin Kleppmann (z-l.pdf",
          "GPT Improving Language Understanding by Generative Pretraining.pdf",
          "Language Models are Few-Shot Learners (GPT3).pdf",
          "Scaling Laws for Neural Language Models.pdf"]

for route in routes:
  currentLoader = PyPDFLoader(
    file_path="/content/drive/MyDrive/Papers/" + route,
    mode = "single",
    extract_images=False
  )
  currentDocument = currentLoader.load()

  currentContent = currentDocument[0].page_content

  documents.append(currentContent)

print("Finished learning")
```

```python
for doc in documents:
  print("===================================== NEW DOC =====================================")
  print(doc.split("\n")[0])
  print(" ")
```
